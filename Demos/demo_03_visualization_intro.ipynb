### Context: Motor Trend Car Road Tests
This example uses a small set of data from the 1970s with road tests from cars.  This is the classic dataset that statisticians have been using for the last 50 years to learn to work with data.  

| Column | Description                              |
| ------ | ---------------------------------------- |
| mpg    | Miles per gallon (fuel efficiency)       |
| cyl    | Number of cylinders                      |
| disp   | Displacement (cu. in.)                   |
| hp     | Gross horsepower                         |
| drat   | Rear axle ratio                          |
| wt     | Weight (1000 lbs)                        |
| qsec   | ¼ mile time                              |
| vs     | Engine type (0 = V-shaped, 1 = straight) |
| am     | Transmission (0 = automatic, 1 = manual) |
| gear   | Number of forward gears                  |
| carb   | Number of carburetors                    |


Your task is to visualize the data using Python functions.


*   **mpg** measures how many miles a car can travel per gallon of fuel, indicating its fuel efficiency.
*   **cyl** indicates the number of cylinders in the engine, which affects power and fuel consumption.
*   **disp** represents the engine displacement in cubic inches, reflecting the overall engine size.
*   **hp** measures the gross horsepower of the engine, showing how powerful the car is.
*   **drat** describes the rear axle ratio, which influences acceleration and driving efficiency.
*   **wt** records the car’s weight in thousands of pounds, with heavier cars typically using more fuel.
*   **qsec** measures the time in seconds required to complete a quarter-mile, indicating acceleration performance.
*   **vs** identifies the engine type, where 0 represents a V-shaped engine and 1 represents a straight engine.
*   **am** indicates the transmission type, where 0 is automatic and 1 is manual.
*   **gear** shows the number of forward gears available in the transmission.
*   **carb** records the number of carburetors supplying fuel to the engine, often associated with performance.

### Import libraries

We will import two libraries
- Pandas, which is like Excel for Python.  It creates 2-dimensional data frames and lets you work with the rows and columns.  
- StatsModels has sample data for use in experimenting with Python

```python
import pandas as pd
import statsmodels.api as sm
```

### Import Sample Data

Import the data that was used in lab 1

```python
mtcars = sm.datasets.get_rdataset("mtcars", "datasets", cache=True).data
df = pd.DataFrame(mtcars)
print(df)
```

```python

df['carmodel'] = df.index
print(df)
```

### Create a Chart

We will create a line chart using Pandas.   Is a line chart the best way to present this data?  What would be a better chart to use?

```python
import matplotlib.pyplot as plt

#Line chart
df1 = df.plot(x='carmodel', y='mpg', kind='line', title="MPG by Car", figsize=(10, 6))
plt.xticks(rotation=90)
plt.show()
```

### Aggregate Data for a Summary Chart

Next we will summarize the data into a new dataframe.   Then create a chart with the aggregated data.

```python
df2 = df['cyl'].value_counts()
print(df2)
```

```python
df2.plot(kind='pie', title='Distribution of Cars by Cylinders')
```
