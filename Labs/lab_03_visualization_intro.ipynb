## Getting data into Pandas

In this case we will load data from the statsmodels.org library

MTCARS is a traditional dataset used for learning.



```python
mtcars = sm.datasets.get_rdataset("mtcars", "datasets", cache=True).data
df = pd.DataFrame(mtcars)
df['carmodel'] = df.index
print(df)
```

## Preview Data

Use the following data dictionary to understand the attributes available

|Attribute | Description |
|---------|--------------|
|	mpg	|Miles/(US) gallon|
|	cyl|	Number of cylinders|
|	disp|	Displacement (cu.in.)|
|hp|	Gross horsepower|
|drat|	Rear axle ratio|
|wt|	Weight (1000 lbs)|
|qsec|	1/4 mile time|
|vs|	Engine (0 = V-shaped, 1 = straight)|
|am|	Transmission (0 = automatic, 1 = manual)|
|gear|	Number of forward gears|


# Create basic charts

- Line chart
- Bar chart
- Histogram
- Box plot
- Scatter plot




```python
#Line chart
df.plot(x='carmodel', y='mpg', kind='line', title="MPG by Car")
```

```python
ax = df.plot(x='carmodel', y='mpg', kind='line', title="MPG by Car")
ax.tick_params(axis='x', rotation=90)
```

```python
#Bar chart
df.plot(x='carmodel', y='mpg', kind='bar', title="MPG by Car")
```

```python
#Histogram
df['mpg'].plot(kind='hist', bins=10, title="MPG Distribution")
```

```python
#Boxplot
df['mpg'].plot(kind='box', title = 'Miles Per Gallon')
```

```python
#Scatter plot
df.plot.scatter(x='mpg',y='hp',c='wt')
# df.plot.scatter(x='mpg', y='hp', c='qsec', s=df['wt']*10)
```

| kind              | Chart type           | When to use it                       |
| ----------------- | -------------------- | ------------------------------------ |
| `line`            | Line chart           | Trends over time or ordered data     |
| `bar`             | Vertical bar chart   | Compare categories                   |
| `barh`            | Horizontal bar chart | Compare categories (long labels)     |
| `hist`            | Histogram            | Distribution of a numeric variable   |
| `box`             | Box plot             | Spread, median, outliers             |
| `kde` / `density` | Density plot         | Smoothed distribution                |
| `area`            | Area chart           | Cumulative trends over time          |
| `pie`             | Pie chart            | Proportions (use sparingly)          |
| `scatter`         | Scatter plot         | Relationship between two variables   |
| `hexbin`          | Hexbin plot          | Dense scatter plots (large datasets) |
